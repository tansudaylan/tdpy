# Practical NumPy Random Arrays

This notebook continues the random-variable examples with reproducible array generation, inspection, visualization, and file round-tripping.

## 1. Setup Notebook Environment
Import NumPy and Matplotlib, then set compact array display options.

## 2. Initialize NumPy Random Generator
Use `numpy.random.default_rng` with a fixed seed so the examples reproduce exactly.

## 3. Generate Uniform Random Arrays
Create one-, two-, and three-dimensional arrays from a bounded uniform distribution.

## 4. Generate Normal Random Arrays
Generate normal samples with a chosen mean and standard deviation.

## 5. Generate Random Integers and Samples
Draw bounded integers and sample distinct values from a finite population.

## 6. Shuffle and Permute Data
Compare in-place shuffling with a returned permutation.

## 7. Inspect Shapes and Summary Statistics
Check each array's shape, dtype, mean, variance, minimum, and maximum.

## 8. Visualize Random Data Distributions
Compare normalized histograms for uniform and normal draws.

## 9. Persist Arrays to Disk
Save arrays to a compressed NumPy archive, load it, and verify round-trip equality.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

import matplotlib.pyplot as plt
import numpy as np

# 1. Setup display options for concise array output.
np.set_printoptions(precision=3, suppress=True, linewidth=100)

# 2. Use a fixed seed so every result is reproducible.
rng = np.random.default_rng(20261001)

# 3. Uniform arrays with explicit shapes and value ranges.
uniform_1d = rng.uniform(-2.0, 3.0, size=8)
uniform_2d = rng.uniform(0.0, 1.0, size=(3, 4))
uniform_3d = rng.uniform(10.0, 20.0, size=(2, 3, 4))
uniform_draws = rng.uniform(-2.0, 3.0, size=5000)

# 4. Normal arrays with configurable location and scale.
normal_1d = rng.normal(loc=2.0, scale=0.5, size=5000)
normal_2d = rng.normal(loc=0.0, scale=1.5, size=(4, 5))

# 5. Bounded integer draws and sampling without replacement.
random_integers = rng.integers(low=0, high=100, size=12)
population = np.arange(100, 120)
random_sample = rng.choice(population, size=6, replace=False)

# 6. Shuffle mutates its input; permutation returns a reordered copy.
shuffle_data = np.arange(10)
rng.shuffle(shuffle_data)
permuted_data = rng.permutation(np.arange(10))

print("Uniform 1D:", uniform_1d)
print("Uniform 2D shape:", uniform_2d.shape)
print("Uniform 3D shape:", uniform_3d.shape)
print("Normal 2D shape:", normal_2d.shape)
print("Random integers:", random_integers)
print("Sample without replacement:", random_sample)
print("In-place shuffle:", shuffle_data)
print("Permutation:", permuted_data)

# 7. Inspect dimensions, storage types, and summary statistics.
summary_arrays = {
    "uniform_1d": uniform_1d,
    "uniform_2d": uniform_2d,
    "uniform_3d": uniform_3d,
    "uniform_draws": uniform_draws,
    "normal_1d": normal_1d,
    "normal_2d": normal_2d,
    "random_integers": random_integers,
    "random_sample": random_sample,
    "shuffle_data": shuffle_data,
    "permuted_data": permuted_data,
}
for name, values in summary_arrays.items():
    print(
        f"{name}: shape={values.shape}, dtype={values.dtype}, "
        f"mean={values.mean():.3f}, variance={values.var():.3f}, "
        f"min={values.min():.3f}, max={values.max():.3f}"
    )

# 8. Compare normalized uniform and normal distributions.
figure, axis = plt.subplots(figsize=(7.0, 4.0), facecolor="white")
axis.set_facecolor("white")
axis.hist(uniform_draws, bins=35, density=True, histtype="step", linewidth=1.8,
          color="#A51C30", label="Uniform(-2, 3)")
axis.hist(normal_1d, bins=35, density=True, histtype="step", linewidth=1.8,
          color="#202020", label="Normal(2, 0.5)")
axis.set(xlabel="Value", ylabel="Probability density",
         title="Uniform and normal random draws")
axis.grid(False)
axis.legend(frameon=True, fancybox=True, framealpha=1.0, facecolor="white",
            edgecolor="black")
plt.show()

# 9. Save and reload arrays in a temporary compressed NumPy archive.
with TemporaryDirectory(prefix="numpy-random-arrays-") as directory:
    archive_path = Path(directory) / "random_arrays.npz"
    print(f"Writing to {archive_path}...")
    np.savez_compressed(archive_path, **summary_arrays)
    print(f"Reading from {archive_path}...")
    with np.load(archive_path) as archive:
        for name, values in summary_arrays.items():
            assert np.array_equal(archive[name], values)
print("Round-trip checks passed.")
```
